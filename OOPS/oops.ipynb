```python
class ComplexNumber:
  def __init__(self, imag, real):
    self.real = real
    self.imag = imag

  def __str__(self):
    return f"{self.real} {self.imag}"

  def conjugate(self):
    print(f"{self.real} + {self.imag}")

```

```python
c1 = ComplexNumber(1,2)
print(c1)
```
